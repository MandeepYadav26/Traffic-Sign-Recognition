### 📦 Step 1.1: Verify Python Version & Core System Info

```python
import sys
import os
import platform

print("=" * 50)
print("🖥️  SYSTEM ENVIRONMENT INFORMATION")
print("=" * 50)
print("Python Version    :", sys.version.split()[0])
print("Platform / OS     :", platform.platform())
print("Working Directory :", os.getcwd())
print("=" * 50)

if sys.version_info >= (3, 8):
    print("✅ Python version is compatible with TensorFlow!")
else:
    print("⚠️ Warning: Python 3.8 or newer is recommended for TensorFlow.")
```

### 📦 Step 1.2: Check and Import Required Libraries
Let's import our deep learning, computer vision, and data science tools.

```python
missing_packages = []

try:
    import numpy as np
    print(f"✅ NumPy Version        : {np.__version__}")
except ImportError:
    missing_packages.append("numpy")

try:
    import matplotlib
    import matplotlib.pyplot as plt
    print(f"✅ Matplotlib Version  : {matplotlib.__version__}")
except ImportError:
    missing_packages.append("matplotlib")

try:
    import cv2
    print(f"✅ OpenCV Version      : {cv2.__version__}")
except ImportError:
    missing_packages.append("opencv-python")

try:
    import sklearn
    print(f"✅ Scikit-Learn Version: {sklearn.__version__}")
except ImportError:
    missing_packages.append("scikit-learn")

try:
    import tensorflow as tf
    print(f"✅ TensorFlow Version  : {tf.__version__}")
except ImportError:
    missing_packages.append("tensorflow")

try:
    import flask
    print(f"✅ Flask Version       : {flask.__version__}")
except ImportError:
    missing_packages.append("flask")

if missing_packages:
    print("\n❌ Some packages are missing:", missing_packages)
    print("Please run in your terminal: pip install " + " ".join(missing_packages))
else:
    print("\n🎉 All essential libraries are installed and ready!")
```

### ⚡ Step 1.3: Check Hardware Acceleration (GPU vs CPU)
Deep learning can train much faster if an NVIDIA GPU with CUDA is present. If you don't have a dedicated GPU, don't worry: our baseline CNN model is lightweight and will train in just a few minutes on standard CPU.

```python
gpus = tf.config.list_physical_devices('GPU')

if gpus:
    print(f"🚀 GPU Detected! Found {len(gpus)} GPU(s):")
    for gpu in gpus:
        print(f"   - {gpu.name}")
    print("Your models will train with GPU hardware acceleration!")
else:
    print("ℹ️ No dedicated GPU detected. The model will run on CPU.")
    print("Don't worry: our 32x32 CNN model is lightweight and trains fast on CPU as well!")
```

### 📁 Step 1.4: Automatically Set Up Project Directories
We create all necessary folders so the rest of the project runs smoothly without missing directory errors.

```python
# Define required folders
folders_to_create = [
    os.path.join("..", "dataset"),
    os.path.join("..", "model"),
    os.path.join("..", "static", "uploads"),
    os.path.join("..", "static", "samples"),
    os.path.join("..", "templates")
]

print("Creating project directories:")
for folder in folders_to_create:
    os.makedirs(folder, exist_ok=True)
    print(f"  [OK] {os.path.abspath(folder)}")

print("\n✅ Project folder hierarchy is set up!")
```

### 🖼️ Step 1.5: Test OpenCV & Matplotlib Image Rendering
Let's create a sample test image (a red circle resembling a speed limit sign) to test our image display pipeline.

```python
# Create a blank 32x32 RGB image (white background)
test_image = np.full((32, 32, 3), 255, dtype=np.uint8)

# Draw a red outer ring (like a speed limit sign)
cv2.circle(test_image, (16, 16), 13, (220, 20, 20), 3)

# Display using Matplotlib
plt.figure(figsize=(3, 3))
plt.imshow(test_image)
plt.title("Test Sign Rendering (32x32)", fontsize=10)
plt.axis('off')
plt.show()

print("✅ Image processing & rendering pipeline works perfectly!")
```
