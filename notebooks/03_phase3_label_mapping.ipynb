### 📁 Step 3.1: Locate Project Root & Import Libraries

```python
import os
import sys
import json
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Handle path when running inside notebooks folder or root
if os.path.basename(os.getcwd()) == "notebooks":
    PROJECT_ROOT = os.path.abspath("..")
else:
    PROJECT_ROOT = os.path.abspath(".")

sys.path.append(PROJECT_ROOT)
from label_utils import CLASS_NAMES, SIGN_CATEGORIES, SIGN_SHAPES, get_full_metadata

print("Project Root :", PROJECT_ROOT)
print(f"Loaded {len(CLASS_NAMES)} class definitions.")
```

### 🏷️ Step 3.2: Inspect Sign Categories & Distribution
Let's see how many traffic signs belong to each functional group.

```python
from collections import Counter

category_counts = Counter(SIGN_CATEGORIES.values())

print("=" * 45)
print("📊 TRAFFIC SIGN FUNCTIONAL CATEGORIES")
print("=" * 45)
for category, count in category_counts.most_common():
    print(f"  • {category:<20} : {count:2d} classes")
print("=" * 45)

# Bar chart of category counts
plt.figure(figsize=(9, 4))
plt.bar(category_counts.keys(), category_counts.values(), color=['#38bdf8', '#f87171', '#4ade80', '#fbbf24', '#a78bfa', '#94a3b8'])
plt.title("Number of Classes per Sign Category (43 Total)", fontsize=12)
plt.xlabel("Category")
plt.ylabel("Number of Classes")
plt.xticks(rotation=20)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 📋 Step 3.3: Full 43-Class Reference Table
Let's list all 43 classes with their IDs, official names, functional categories, and shapes.

```python
print(f"{'ID':<4} | {'Traffic Sign Name':<38} | {'Category':<18} | {'Shape':<15}")
print("-" * 85)
for cid in range(43):
    meta = get_full_metadata(cid)
    print(f"{meta['class_id']:<4} | {meta['name']:<38} | {meta['category']:<18} | {meta['shape']:<15}")
```

### 🖼️ Step 3.4: Visual Gallery of Key Sign Categories
Let's load one real image from the dataset for distinct sign geometries to inspect how visual features correspond to classes:
- **Octagon**: Stop (Class 14)
- **Inverted Triangle**: Yield (Class 13)
- **Diamond**: Priority Road (Class 12)
- **Red Ring Circle**: Speed Limit 50 (Class 2)
- **Blue Circle**: Roundabout (Class 40)
- **Warning Triangle**: Traffic Signals (Class 26)

```python
dataset_path = os.path.join(PROJECT_ROOT, "dataset", "gtsrb")
key_classes = [14, 13, 12, 2, 40, 26]

if os.path.exists(dataset_path):
    plt.figure(figsize=(12, 7))
    for i, cid in enumerate(key_classes):
        folder = os.path.join(dataset_path, str(cid))
        if os.path.exists(folder) and len(os.listdir(folder)) > 0:
            sample_file = os.path.join(folder, os.listdir(folder)[0])
            img = cv2.imread(sample_file)
            img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            
            plt.subplot(2, 3, i + 1)
            plt.imshow(img_rgb)
            meta = get_full_metadata(cid)
            plt.title(f"[{cid}] {meta['name']}\n({meta['shape']}, {meta['category']})", fontsize=10)
            plt.axis('off')
    plt.tight_layout()
    plt.show()
else:
    print("Dataset folder not found. Please run Phase 2 notebook to download images first.")
```

### 💾 Step 3.5: Verify and Save Metadata Files
We confirm that `class_labels.json` and `class_metadata.json` are generated and can be loaded cleanly by any script or web app.

```python
labels_file = os.path.join(PROJECT_ROOT, "class_labels.json")
metadata_file = os.path.join(PROJECT_ROOT, "class_metadata.json")

# Verify files
print(f"Checking {labels_file} ...")
with open(labels_file, "r") as f:
    labels_data = json.load(f)
print(f"  [OK] Loaded {len(labels_data)} labels.")

print(f"Checking {metadata_file} ...")
with open(metadata_file, "r") as f:
    meta_data = json.load(f)
print(f"  [OK] Loaded {len(meta_data)} metadata entries with shapes & categories.")

print("\n✅ Phase 3 Complete: Metadata mapping is verified and ready for EDA in Phase 4!")
```
