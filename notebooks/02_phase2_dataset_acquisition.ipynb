### 📁 Step 2.1: Locate Project Root & Import Tools

```python
import os
import sys
import zipfile
import urllib.request

# Determine project root path (handles running from root or notebooks folder)
if os.path.basename(os.getcwd()) == "notebooks":
    PROJECT_ROOT = os.path.abspath("..")
else:
    PROJECT_ROOT = os.path.abspath(".")

DATASET_DIR = os.path.join(PROJECT_ROOT, "dataset")
DATA_PATH = os.path.join(DATASET_DIR, "gtsrb")
ZIP_FILE_PATH = os.path.join(DATASET_DIR, "gtsrb.zip")
DOWNLOAD_URL = "https://cdn.cs50.net/ai/2020/x/projects/5/gtsrb.zip"

print("Project Root :", PROJECT_ROOT)
print("Dataset Path :", DATA_PATH)
```

### 📥 Step 2.2: Download the GTSRB Archive
If the dataset is already extracted, this step will detect it and avoid re-downloading.

```python
def download_progress(count, block_size, total_size):
    """Displays download progress in the notebook."""
    downloaded = count * block_size
    if total_size > 0:
        percent = min(100, downloaded * 100 / total_size)
        mb_down = downloaded / (1024 * 1024)
        mb_total = total_size / (1024 * 1024)
        sys.stdout.write(f"\r📥 Downloading: {percent:5.1f}% ({mb_down:.1f} MB of {mb_total:.1f} MB)")
        sys.stdout.flush()

# Check if dataset already exists
if os.path.exists(DATA_PATH) and len(os.listdir(DATA_PATH)) >= 43:
    print(f"✅ Dataset is already downloaded and present at:\n   {DATA_PATH}")
else:
    os.makedirs(DATASET_DIR, exist_ok=True)
    print(f"Starting download from: {DOWNLOAD_URL}")
    print("This is a ~200MB file and may take 1-3 minutes depending on your internet connection...\n")
    urllib.request.urlretrieve(DOWNLOAD_URL, ZIP_FILE_PATH, reporthook=download_progress)
    print("\n\n✅ Download complete! Zip file saved.")
```

### 📦 Step 2.3: Extract the Archive
We extract all contents into `dataset/` and then delete the ZIP file to free up disk space.

```python
if os.path.exists(ZIP_FILE_PATH):
    print("Extracting dataset archive (this takes about 10-20 seconds)...")
    with zipfile.ZipFile(ZIP_FILE_PATH, 'r') as zip_ref:
        zip_ref.extractall(DATASET_DIR)
    print("Extraction finished!")
    
    # Remove the ZIP archive to save ~200MB disk space
    os.remove(ZIP_FILE_PATH)
    print(f"Removed temporary file: {ZIP_FILE_PATH}")
else:
    print("Zip file already extracted or not present.")
```

### 🔍 Step 2.4: Verify the 43 Classes & Count Images
Let's confirm that folders `0` through `42` exist and count how many images are inside.

```python
if not os.path.exists(DATA_PATH):
    print(f"❌ Error: Dataset directory not found at: {DATA_PATH}")
else:
    folders = sorted([f for f in os.listdir(DATA_PATH) if os.path.isdir(os.path.join(DATA_PATH, f))], key=lambda x: int(x) if x.isdigit() else 999)
    print(f"Found {len(folders)} class folders (Expected: 43 folders numbered 0 to 42).")
    
    total_images = 0
    counts = {}
    for folder in folders:
        folder_path = os.path.join(DATA_PATH, folder)
        num_files = len(os.listdir(folder_path))
        counts[folder] = num_files
        total_images += num_files
    
    print(f"\n📊 Dataset Statistics:")
    print(f"  • Total Classes : {len(folders)}")
    print(f"  • Total Images  : {total_images:,}")
    print(f"  • Minimum Images in a Class : {min(counts.values())} (Class {min(counts, key=counts.get)})")
    print(f"  • Maximum Images in a Class : {max(counts.values())} (Class {max(counts, key=counts.get)})")
    print("\nFirst 5 class folder image counts:")
    for f in folders[:5]:
        print(f"   Folder '{f}': {counts[f]} images")
    print("   ...")
    
    if len(folders) == 43:
        print("\n🎉 Verification PASSED: All 43 traffic sign classes are present!")
    else:
        print(f"\n⚠️ Warning: Expected 43 classes, but found {len(folders)}.")
```
